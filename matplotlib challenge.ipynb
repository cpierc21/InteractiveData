```python
import pandas as pd
from matplotlib import pyplot as plt
```

```python
## I kinda want to learn violins because a lot of my work is in sports statistics.

## So, a lot of data repeats... a lot. So to see visibly, the frequency a particular data point shows up is valuable.

## And any type of frequency charts are good, because team scores, individual scores, or individual stat lines...
## Seeing frequency is... key.
```

```python
pew = pd.read_csv('cces_example_data.csv')
pew.head()
```

```python
fig1, ax1=plt.subplots()
ax1.hist(pew['age'],bins=90)
ax1.set_xlabel("Age Histogram")
ax1.set_ylabel("Political Identity")
```

```python
freq = pew["Evaluation"].value_counts(dropna=False)

fig2, ax2 = plt.subplots()
freq.plot(kind="bar")

ax2.set_xlabel("Evaluation")
ax2.set_ylabel("Frequency")
ax2.set_title("Frequencies of Responses in Evaluation")

plt.tight_layout()
plt.show()
```

```python
fig3, ax3 = plt.subplots()

ax3.scatter(pew["ConMediaUse"], pew["ImmiPhobia"])

ax3.set_xlabel("Conservative Media Usage")
ax3.set_ylabel("Immigrant Phobia")
ax3.set_title("Does Conservative Media Use Correlate to Immigrant Phobia")

plt.tight_layout()
plt.show()
```
